#### O que é um modelo computacional?

Vimos o conceito de arquitetura, organização e implementação. Um modelo computacional é um conceito _mais abstrato_ do que arquitetura. O modelo inclui premissas fundamentais comuns a um conjunto de arquiteturas. Então...

- **Modelo computacional:** Conjunto de regras que são seguidas em um projeto de uma classe de arquiteturas (Von Neumann). Existem dois aspectos do lado operacional:

    - Um trata das possíveis ações que uma arquitetura pode executar, quando elas podem ser executadas e como os dados são armazenados e recuperados. 
    
    - Outro lado permite a análise de algorítmos e permite estimar quão eficiente um algorítmo poderá ser executado

        ![algorit.png](algorit.png)

- **Arquitetura:** É uq o programador em linguagem montadora vê no computador. Então se dois computadores utilizam a mesma linguagem montadora, eles têm a mesma arquitetura (arm, x86, risc-V, vax-11, 8085)

- **Organização:** São coisas específicas de cada arquitetura. Pode ser memória cache por exemplo, placa de vídeo integrada ou não (i5, i7, I/O, qtd de núcleos)

- **Implementação:** Aqui são especificadas as características mais específias ainda, em um modelo de hardware. (número de transistores, velocidade de clock)

# Visão Operacional

Um modelo computacional deverá incluir:

- Ações básicas que podem ser executadas
- Maneira como os dados são armazenados e acessados _(mecanismo de dados)_
- Maneira com as ações são executadas e controladas _(mecanismo de controle)_
- Modo de comunicação entre as unidades de processamento _(quando existir 1+)_
- Macanismo de sincronização entre as unidades de processamento _(quando existir 1+)_

No modelo de von Neumann, as características são:

- As instruções são do tipo multiplique este dado por aquele e coloque o resultado em
tal lugar.
- Os dados são armazenados juntamente com o programa num espaço linear de armazenamento. Esse espaço contém cédulas com endereço ́unico.
- As instruções são executadas, uma após a outra, e são buscadas na memória sequencialmente.

As principais críticas a esse modelo são:

- A existência de uma ligação (UCP à memória) que ́e sobrecarregada (dados e instruções passam por ela, os data bus e  instruction bus) e facilmente se torna um gargalo.
- A existência de “efeitos colaterais”, pois a memória de dados se mistura com a memória de programas.

## Classificações de Flynn

Foi criada por Flynn em 1966

- **SISD (_Single Instruction Single Data):** É o modelo von Neumann convencional. Nesse modelo, existe sempre uma ́unica instrução agindo sobre um  ́unico dado num instante de tempo qualquer. **UMA instrução agindo sobre UM dado**

- **SIMD (_Single Instruction Multiple Data):** Aqui uma ́unica instrução age sobre um conjunto de dados simultaneamente. Pode ser uma multiplicação de um vetor por um escalar. Nessa classe incluimos os processadores de arranjo (“array”) e os processadores de vetores (“vector”). **UMA instrução agindo sobre VÁRIOS dados**

- **MIMD (_Multiple Instruction Multiple Data):** Neste caso, estão os computadores onde múltiplas instruções agem sobre múltiplos dados. A maioria das pesquisas em arquitetura paralela se concentram hoje nesta categoria. **VÁRIAS instruções agindo sobre VÁRIOS dados**

- MISD pode ser considerado um pipeline
