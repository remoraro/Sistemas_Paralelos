#### Arithmetic Unit
A soma em binário de 0 + 1 = 1 + 0 = 1. Mas se somarmos 1 + 1 não da 2, da 10. Como podemos ver, *precisamos adicionar um bit* para conseguirmos somar os números, ou se não temos um ***overflow***. Esse bit é chamado de ***carry***. Um **FULL ADDER** recebe **3 bits** para fazer a soma, 2 bits para somar e um carry da soma anterior (se houver) e devolve **2 bits**, a soma em si e o carry se houver.

#### Logic Unit
Aqui são aplicadas mais gates de _True_ e _False_, para ver se um número é zero, se é negativo etc

Para representar a _ALU_ usamos um formato de "V" meio estranho, mas que é bem intuitivo

![alu.png](alu.png)

Exemplos dessa representação: Se queremos fazer 1 - 1, no input A colocamos 1, no input B colocamos 1, no operation code colocamos o código para subtração. A resposta vai dar 0 e a flag de zero vai ser acessa
Se queremos comparar se 10 > 5, nós subtraímos esses 2 números e vemos se a flag de negative é acessa. Se for acessa, 10 < 5, se continuar apagada 10 > 5

# Aula 7

Aqui vamos falar da **CPU**. Olhando para a imagem da ALU podemos compreender a imagem abaixo de um **Ciclo de Instruções:**

![assembly.png](assembly.png)

Da pra ver a semelhança com a linguagem **Assembly** né? Você fala exatamente o que fazer para o computador. Vamos entender melhor esse ciclo!

* Temos um **Registrador para o Endereço das Instruções**, nele marcamos em qual passo estamos, no 0, no 1, no 2, no 3 etc
* Também temos um **Registrador para a Instrução em si**, que é apontado por esse anterior

Agora, nas instruções em si, podemos ter:
- **Fetch (F):**  Pega a instrução da memória, como por exemplo pega 00101110 do endereço 0000 da RAM
- **Decode (D):**  Interpreta a instrução, como por exemplo pega 0010 (LOAD_A)
- **Execute (E):**  Usa os transistores com as gates, como por exemplo aloca o número do endereço 1110 no registrador A
- **Writeback (W):**  Salva o resultado na memória, como por exemplo atualiza o registrador para a próxima instrução

Tudo isso é feito na **Control Unit**, que trabalha de forma conjunta com a ALU para realizar as operações. Todas essas operações são feitas "de uma vez" a cada **Ciclo de Clock**. O esquema completa fica assim:
