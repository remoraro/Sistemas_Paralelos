# Aula 1

No início de tudo, um **_computador_** era alguém que **_computava!_** Essa pessoa fazia contas usando umas máquinas para facilitar os cálculos. 

Após isso vieram máquinas **enormes** que faziam esses cálculos automaticamente. Para ter noção, era possível 3 adições por segundo e uma divisão demorava 15. 

Curiosidade: Foi ai que surgiu o termo "bug", porque o computador rodava em papel perfurado e quando um inseto entrava atrapalhava o código

# Aula 2

Antes dos **transistores** eram usados **tubos com vácuos**, que tinham um mecanismo parecido, mas eram muito problemáticos e frágeis.

![imagecopy.png](imagecopy.png) No transistor tem uma `entrada`, uma `saída` e um `controle`.

O *`Controle`* pode se tornar **_condutor ou isolante_**, **_permitindo a passagem de corrente ou não_**

Atualmente, os transistores estão na escala de *nano* e podem trocar de estado *milhões de vezes por segundo*



Curiosidade: Esse desenvolvimento aconteceu entre os vales de São Francisco e de São José, na Califórnia. Como os transistores usavam muito silício, esse lugar ficou conhecido como Vale do Silício

#### Por que binário?

Um sistema binário como o __transistor__ tende a ter menos erros devído a *ruído* do que sistemas com mais opções.

A **`Álgebra Booleana`** já estava bem desenvolvida também, combinando muito com sistemas binários. A partir dai vieram as portas lógicas:

* A porta `AND`, por exemplo, são 2 transistores em série
