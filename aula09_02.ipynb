## O que é um Bandwidth?

A Largura de Banda (_Bandwidth_) de uma determinada organização de memória é definida como o número
médio de palavras por unidade de tempo que a organização pode fornecer. 

É simplesmente a *capacidade máxima de transmissão de dados*. Uma analogia boa é de uma estrada ou um cano; quanto ***maior a largura de banda***, ***mais dados conseguem trafegar*** simultaneamente

A Largura de Banda segue a fórmula:

 $\text{Largura de Banda} = \text{Largura do Barramento (bits)} \times \text{Frequência (Hz)} \times \text{Transferências por Ciclo}$

#### Conceitos: 
- **Barramento:** Sistema de comunicação que transfere dados entre os componentes internos de um computador. A **largura** deste barramento (ex: 32 bits ou 64 bits) dita quantos bits podem ser enviados de uma vez. (Existem diferentes barramentos, de _addres_, de _data_ e _write/read_).
- **Frquência:** É a frequência do processador, quantos clock cycles ele faz por segndo.
- **Transferências por Ciclo:** É onde divide em **SDR** e **DDR**. Como vimos antes, _SDR_ atua apenas na _subida_ do Clock Cycle e _DDR_ atua na _subida e descida_.

Normalmente dividimos esse valor por **8** para transformarmos em ***bytes***. 

#### Exemplo Prático: Memória RAM DDR4

Uma memória **DDR4-3200MHZ 64Bits** tem uma frequência real de 1600 (é anunciado como o dobro pq é DDR). Então usando a fórmula de antes  $\text{Largura de Banda} = \frac{(64Bits * 1600MHZ * 2)}{8}= 25,6GB/s$

## Processamento Vetorial

Existem funções que:
- levam *um* array para *array* $f_1: V→V$ (exemplo, raiz de um vetor).
- levam *um array* para *escalar* $f_2: V→S$ (exemplo, amplitude de um vetor).
- levam *dois* arrays para *arrays* $f_3: V×V$ (exemplo, soma de vetores)
- levam *um* vetor e *um* escalar para *um* vetor $f_4: V×S$ (exemplo, dobrar um vetor)
