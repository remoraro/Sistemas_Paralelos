# Problemas da Computação Paralela

- **Número de processadores:** Supostamente, quanto mais processadores, melhor seria o desempenho da máquina. _(aqui o prof ficou falando de cryptomoedas e como para mineração é melhor ter mais processadores)_.

- **Comunicação:** Os vários núcleos precisam se comunicar para trocar informação, mas essa comunicação pode ser lenta. Um problema é no fluxo de dados e a _porcentagem_ de comunicação em relação ao tempo de processamento.

  - _Handshake:_ Comunicação síncrona, normalmente mais rápida que assíncrona

- **Controle de Tarefas:** O tamanho de cada tarefa (granularidade grossa ou fina) tem desempenhos diferentes e controlar a execução das tarefas em vários cores. O sistema operacional pode influenciar no desempenho também.
    - _Mestre/Escravo:_ Um dos processadores fica encarregado de pegar as tarefas e distribuir para os outros cores disponíveis

- **Natureza das Aplicações:** O tipo de algoritmo executado interfere no desempenho. Podemos tentar unificar e criar uma máquina paralela boa para tudo, mas máquinas paralelas normalmente são melhores em certos casos e piores em outros.

- **Confiabilidade:** É interessante se algum componente parar de funcionar ou der problema, que o programa não quebre/pare totalmente, mas sim que continue funcionando com certo prejuízo de desempenho. Ainda não há um modelo universalmente aceito para tolerância a falhas
