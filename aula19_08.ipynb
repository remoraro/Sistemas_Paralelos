# **Pipeline:** 
Funciona como uma ***linha de montagem*** com **uma instrução após a outra**

### Como Pipelining mudam os cycles:

Para entender a mudança, vamos ver como uma CPU processa uma única instrução. Ela é quebrada em **4 passos:**

- **Fetch (F):**  Pega a intrução da memória
- **Decode (D):**  Interpreta a instrução
- **Execute (E):**  Usa os transistores com as gates
- **Writeback (W):**  Salva o resultado na memória

*OBS: Sem o pipeline, cada uma dessas coisas demoraria um clock cycle, com ele faz todas ao mesmo tempo*

Primeiro vem um comando, que leva a um `Estado`, e o sistema `Trava` nesse estado (armazena os bits no registro), não permetindo progredir e nem regredir. No próximo `Ciclo` esses bits vão para o próximo estado. 

Tudo é coordenado pelo `Clock` interno.

 
###### (sup = num do dado, sub = num da instrução)

Enquanto um dado passa para o **State 2** chega outro dado que vai para o **State 1**, isso para cada *clock cycle*.



Essas ondas verticais representam as voltagens: 
- Na **subida** (latch) os registradores / transistores **recebem os dados**; 
- No **meio** do cycle **são aplicadas as gates**, nele tem que aplicar tudo e estabilizar, junto com o delay;
- Na **descida** os registradores / transistores jogam seus dados para a próxima sequência e o estado é resetado para receber um novo dado (SDR) ou faz a mesma coisa que a subida (DDR)
- No **fim** do cycle **também são aplicadas as gates**

#### Conceitos:
* Single Data Rate (SDR): The chip only transfers data or moves bits on the rising edge. The falling edge is just used to reset the wave.

* Double Data Rate (DDR): The chip transfers data on both the rising edge and the falling edge. This means the computer does work twice as fast without needing to increase the actual clock speed. This is exactly how your computer's DDR RAM works!

Dúvidas:
- Clock cycle vs Instruction Cycle

res: Um Instruction Cycle são as sequências de execuções para fazer uma instrução. É composto de vários Clock Cycles e é a conta (num_de_clock_cycle * período_clock). Analogia: Clock Cycle = Batida de pernas, Instruction Cycle = correr uma volta inteira na pista

- Qual é o tempo mínimo do computador? 

res: É o Clock Cycle


- Clock cycle é um número fixo ou depende de algo?

res: É um número fixo que depende do clock speed. Mas uma instrução *em código de máquina* como add é especificada pelo fabricante quantos clock cycles ela leva. Exemplo: um add (instrução) dura 3 clock cycles, 3 add duram 9 clock cycles


- Clock speed é a mesma coisa que a frequencia do computador? É um número fixo?

res: sim e depende (em modo ocioso pode diminuir para salvar energia), quantidade de clock cycles por segundo


- Onde são aplicadas as gates? (OR, NOT, AND, etc)

res: Nas partes estáveis
