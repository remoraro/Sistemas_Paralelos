**Arquitetura de computadores** é a ***maneira*** como o computador é **organizado**, com CPU, HD/SSD, RAM, GPU, Conjunto de instruções, num de bits, tipo de endereçamento de memória etc

Um computador (na arquitetura de Von Neumman) está dividido em `CPU` <-> `Entrada/Saída` <-> `Memória`

![alt text](image.png)

Na `CPU`:
* **Unidade de Controle:** Organiza e gerencia a execução das instruções de forma correta
* **Aritmetic Logic Unit:** Realiza as operações/portas lógicas
* **Registradores:** É uma memória de baixo nível para armazenar dados imediatos e instruções da CPU

Na memória, tem `Primária` e `Secundária`, o que hoje chamamos de `RAM` e `Armazenamento`. 
* **A RAM:** se comunica com o processador diretamente para a execução de programas, guardando dados de maneira *temporária*. 
* **O Armazenamento:** armazena dados de maneira *permanente* que podem ser acessados e "jogados" para a primeira quando necessário.

A arquitetura x86 é uma família de conjuntos de instruções de processadores. Ele tem:
* **Compatibilidade Retroativa:** Um programa escrito em uma versão anterior (como 8086) consegue rodar em uma versão moderna (como Intel Core / Ryzen)
* **Complexidade CISC:** CISC (Complex Instruction Set Computer) é um design que com *uma instrução (de comprimento variável)* consegue dar *vários comandos*, a CPU consegue *acessar a memória diretamente* e tem *mais clocks cycles* (obs: uma instrução pode levar vários clock cyles para ser concluída)
* **Evolução de Bits:** Começou em 16bits (que significa que consegue registrar 16bits de uma vez), depois 32 bits e agora 64bits. Então o nome "completo" da arquitetura fica x86-64

_Resumo: clock = timer interno, cycle = tick unico desse timer. Clock cycle é o tempo que leva para executar uma ação, um computador com 3GHz repete um cycle 3 bilhões de vezes por segundo_

Um único clock cycle consegue fazer *parte* de uma tarefa grande, ou *várias* tarefas pequenas


**Organização de Computadores** se baseia na *estrutura interna* de um computador, o hardware. Ele pode ter RAM dinâmica ou estática, DDR4 ou DDR5, barramentos (conexão entre as partes), dispositivos de entrada/saída, diferentes pipelines. ***Uma mesma arquitetura pode ter diversas organizações***

Uma organização pode ter várias **implementações**. A implementação é bem autoexplicativa, são os componentes *específicos* que são escolhidos. Então em uma arquitetura que tem uma CPU a implementação pode ser a quantidade e tamanho de transistores, etc.
