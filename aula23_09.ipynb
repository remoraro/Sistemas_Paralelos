### O que é um compilador? Um interpretador? Um Montador?

![compilador.png](compilador.png)

Um `Compilador` pega uma linguagem de **alto nível** para **baixo nível**. Como por exemplo, de ***Python*** para ***Assembly***

Um `Interpretador`pega uma linguagem de **alto nível** para uma **intermediária**. Como por exemplo, de ***Java*** para ***bytecode***. É literalmente um passo intermediário nesse processo.

Um `Montador` pega a linguagem em ***Assembly*** para ***código de máquina***

### Diferença entre linguagem Compilada e Interpretada:

Em uma linguagem `Compilada`, como **C**, é criado um arquivo executável (gcc -o hello hello.c) para o código de máquina e depois ele é executado ***de uma vez***.

Em uma linguagem `Interpretada`, como **Java/Python**, cada linha é ***traduzida e executada por vez***

###### Ainda existe o `JIT` _(Just In Time)_, que é uma mistura das 2 que não vamos estudar
