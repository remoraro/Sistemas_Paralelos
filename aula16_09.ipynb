Foi percebido que muita coisa se baseia em **cálculos vetoriais**. Uma pessoa importante que percebeu isso foi _Seymour Cray_, que criou a Cray (empresa) e criou o **Cray-1** (computador).

Além dessas funções, existem **funções booleanas** como a _compare_ e a _compress_, que aplica uma **máscara vetorial** em um vetor.
Um exemplo é a máscara $M = (1, 1, 0, 1, 0, 0, 0, 1).$ é aplicada em $A = (1, 2, 4, 8)$ e
$B = (3, 5, 6, 7)$, que retorna $C = (1, 2, 3, 4, 5, 6, 7, 8, 9)$

Um código vetorial, pode acessar a _memória a memória_ ou _registrador a registrador_. 
- Na primeira você, para cada cálculo, você pega o número da memória e faz a operação nele. 
- Na segunda você trás todos os números que você quer usar para os registradores e depois acessa eles.

O **Cray** usava o segundo tipo, por isso ele tinha um monte de registradores e por isso ele derreteu da primeira vez.
Em um computador vetorial moderno, uma instrução vetorial envolve os seguintes campos:
- O código do operando.
- O endereço base e o “offset”.
- O incremento do endereço.
- O comprimento do vetor a ser utilizado.

Como exemplo de operação vetorial, considere o código FORTRAN abaixo:

```python
DO i = 1, n
    A[i] = B[i] + C[i]
    B[i] = 2 * A[i]
END DO
```

Aqui da pra ver que precisamos calcular $A[i]$ com o $B[i]$ antigo, se por acaso $B[i]$ fosse calculado antes, teríamos um calculo errado.

Então **não podemos** rodar essas 2 linhas paralelamente, por conta dessa ***racing condition*** (importa quem acaba primeiro e não temos controle disso)

Em um processador vetorial, ficaria assim:

```python
TEMP(1:N) = A(2:N+1)
A(1:N) = B(1:N) + C(1:N)
B(1:N) = 2*TEMP(1:N)
```

O primeiro código **perde muito tempo** no controle do loop, se tornando ineficiente. Por outro lado, computadores que conseguem fazer **cálculos vetorias**, como o Cray-1, até perdem tempo até prepararem os registradores (_setup time_), mas são muito mais eficientes!
