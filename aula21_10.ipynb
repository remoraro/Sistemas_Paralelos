# Granularidade

É o conceito de como uma _tarefa_ é distribuída entre vários processadores.

A ***Granularidade Grossa*** aloca grandes processos (_grains_) para os processadores

A ***Granularidade Fina*** aloca pequenos processos (_grains_) para os processadores. Essas máquinas são linearmente expansíveis porque possuem um alto nível de paralelismo, com uma boa divisão de trabalho. Quando há `complexidade procedural` há dificuldade de paralelizar o problema e obter um speedup real.

Uma máquina de granularidade fina conhecida é o _dataflow_

# Computação Paralela

Tem 3 objetivos principais:

- Executar programas lentos com mais rapidez em supercomputadores
- Executar programas com mais rapidez em computadores comuns
- Aumentar a produtividade de programadores quando escrevem códigos para aplicações intrinsicamente paralelas
