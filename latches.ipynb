## O que é um *Latch* ?

É um **circuito** que consegue **lembrar/armazenar** um bit de ***informação***

Vamos por partes, primeiro vamos ver esse circuito:

![lor.png](lor.png)

Nele temos uma porta `OR`, então vemos que se qualquer um dos inputs for *1* a saída será *1* para sempre. *Como se o circuito lembrasse que recebeu um 1 de input*

Agora vamos para o circuito com a porta `AND`. Nele, se qualquer input for *0* a saída será *0* para sempre.

![land.png](land.png)

_(OBS: Note que essa operação de bits não é instantânea, demora um tempo minúsculo mas demora um tempinho)_

Combinando os 2 circuitos temos o seguinte:

![circuito.png](circuito.png)

Note que em cima temos a porta `OR`, então se o **_input_** for `1` o **_output_** será `1` ***e o input de cima não vai conseguir mudar mais nada***

Mas se o ***input*** de baixo for `1` a porta `AND` mais para a direita vai receber `0`. Então o ***output*** será 0, ***resetando o circuito***

## ***Esse circuito consegue lembrar 1 bit de informação, contendo um SET e um RESET***

![latch.png](latch.png)

Mas com esse modelo de **Latch** temos um problema, podemos criar **estados inválidos** se o SET e o RESET forem iguais a 1. 

Para resolver isso vamos criar um modelo com um input `DATA` que ***recebe o bit*** e um input `WRITE/ENABLE` que ***permite o estado ser registrado ou não***

![write.png](write.png)

Aqui **só é possível mudar o estado** se o `WRITE/ENABLE` estiver **ligado**. O estado **permanece** após a mudança, ***por isso é um Latch***

![enable.png](enable.png)

Mas, somente **1 bit de informação** não é tão útil. Então nós **combinamos** vários *Latches* para criar algo significativo, olhe esse exemplo de **número de 8bits**:

![92.png](92.png)

E sabe como é chamado um conjunto de *Latches*?
### Um REGISTRADOR!

Então um registrador de **8bits** tem **8 *Latches***, um de **64bits** tem **64 *Latches***. Mas ficaria muito ineficiente armazenar um do lado do outro como na imagem. Então eles são organizados em **MATRIZES**

![matriz.png](matriz.png)

Ficou um pouco complicado então vamos com calma...

Cada *bit* vai ter um:
- **In/Out:** Aqui vai ser a nossa entrada e a nossa saída de data, **quando quisermos ler ou escrever um estado a informação vai vir por aqui**
- **Write/Enable:** Aqui vai estar ativado se queremos **escrever** no bit
- **Read/Enable:** Aqui vai estar ativado se queremos **ler** o estado do bit

![inout.png](inout.png)

Mas acessar um **bit específico** usamos um **endereço**. Ele vai ligar *uma das quatro saídas* (00 saída 0, 01 saída 1, 10 saída 2, 11 saída 3) utilizando um **decoder**. Para isso funcionar certinho temos que adicionar uma porta `AND` em cada bit conectando a linha e a coluna, **para ser ativado somente quando os 2 estiverem ligados**.

![andmatriz.png](andmatriz.png) ![decoder.png](decoder.png)

Como queremos acessar **somente 1 bit por vez** e **somente esse bit vai estar acessível** podemos ***conectar todos os bits em 3 linhas para Read, Write e In/Out***

![canto.png](canto.png)

![byte.png](byte.png)

### E como isso se torna ***memória***? Se vermos **bytes como sequências de bits** podemos ver uma **lista de bytes como memória**

![bytefinal.png](bytefinal.png)

Aqui temos algumas coisas para explicar:
- **Address Bus:** É o *endereço* de onde o byte está guardado
- **Write/Read:** Mesmo que antes, permite ler e escrever bytes
- **Data Bus:** É onde a sequência de bits (o byte) vai no Read/Write

Neste exemplo temos um **Address Bus de 4 bits que controla 16 bits individuais (2^4)**, mas com **8 bits de Address Bus conseguimos controlar 256 bits individuais (2^8)**

Por isso um computador com um **Address de 32 bits** consegue armazenar **2^32** que dá 4 Giga-Bytes!

### Um nome meio estranho para isso, como **conseguimos acessar qualquer byte aleatório só dando o endereço dele**, é **Random Acess Memory** a **Memória RAM!**
